# llama-3

```python
import json
from pathlib import Path
import pandas as pd

# Đường dẫn thư mục chứa file JSON
folder = Path(r"D:\Projects\paper-CT\Maternal-Mental-Health\mental_health_multiagent\datasets\chat_logs\full_option_8k\llama-3.3-70b-versatile")

data = []

for i, file in enumerate(folder.glob("chat_*.json")):
    if i >= 500:
        break
    with open(file, 'r', encoding='utf-8') as f:
        data.append(json.load(f))

print(f"Đã load {len(data)} mẫu")
```

```python
def get_full_conversation(conv_list):
    return " ".join([turn["content"] for turn in conv_list if turn["role"] in ["assistant", "patient"]])

# Tạo cột mới
for sample in data:
    sample["full_text"] = get_full_conversation(sample["conversation"])
```

```python
import textstat

results = []

for i, sample in enumerate(data):
    text = sample["full_text"]
    
    fre = textstat.flesch_reading_ease(text)
    fkg = textstat.flesch_kincaid_grade(text)
    gfi = textstat.gunning_fog(text)
    
    results.append({
        "id": i,
        "model": sample["metadata"]["assistant_model"],
        "FRE": fre,
        "FKG": fkg,
        "GFI": gfi
    })

df_read = pd.DataFrame(results)
```

```python
print(df_read.groupby("model")[["FRE", "FKG", "GFI"]].mean().round(2))
```

```python
from bert_score import score
import torch

# Lấy diagnosis làm reference
refs = [sample["diagnosis"] for sample in data]
cands = [sample["full_text"] for sample in data]

# Tính BERTScore
P, R, F1 = score(cands, refs, lang="en", verbose=True)
bertscore_f1 = (F1 * 100).tolist()  # chuyển sang %

# Gán lại
for i, score in enumerate(bertscore_f1):
    results[i]["BERTScore"] = round(score, 2)

print(df_read.groupby("model")["BERTScore"].mean())
```
